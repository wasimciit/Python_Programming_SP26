# Dictionary
- Dictionaries are used to store data values in key:value pairs.

- A dictionary is a collection which is ordered*, changeable and do not allow duplicates.
- Dictionaries are written with curly brackets, and have keys and values:

```python
# creat a dicionary
A={"Class":2,"Age":3.4,"Name":"Ali"}
print(type(A))
```

```python
print(A)
```

```python
# create and print a dictionary
dict1={"Names":("Ali","Baber",'Zahid'),"Age":{23,24,25},'Marital':['Yes',"No",'Divorced']}
print(dict1)
```

```python
# create a dictionary with duplicate values
dict1={"Names":["Ali","Baber",'Zahid'],"Names":["Dilawar","Hamad","Aleena"],"Age":[23,24,25],'Marital':['Yes',"No",'Divorced']}
print(dict1)
```

```python
# len function to find how many items a dictionary has
print(len(dict1))
```

```python
# type function
print(type(dict1))
```

```python
x="Ali"
print(type(x))
y=3
print(type(y))
```

```python
# The values in dictionary items can be of any data type
dict2={"Names":["Ali","Baber",'Zahid'],"Age":23,'Marital':('Yes',"No",'Divorced'),"Marks":{9,0,8.7}}
dict2
```

```python
# The dict() constructor
dict3=dict(({"Names":["Ali","Baber",'Zahid'],"Age":[23,24,25],'Marital':['Yes',"No",'Divorced']}))
print(dict3)
print(type(dict3))
```

```python
dict4={"Names":["Ali","Baber",'Zahid'],"Age":[23,24,25],'Marital':['Yes',"No",'Divorced']}
dict5=dict({'Ali':'Name',"age":23})
print(type(dict4))
print(type(dict5))
```
