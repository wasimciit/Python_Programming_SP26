A **tuple** in Python is an ordered, immutable collection of items. This means that once a tuple is created, you cannot change its elements (add, remove, or modify).

Key characteristics of tuples:
*   **Ordered**: The items have a defined order, and that order will not change.
*   **Immutable**: You cannot change elements of a tuple once it's assigned. This makes them suitable for data that should not be altered.
*   **Allows duplicates**: Tuples can have duplicate members.


Tuples are defined by enclosing the elements in parentheses `()` and separating them with commas.

```python
# Creating tuples

num=(20,30,34,40,40.5,"Ali",'Ali',True)
print(num)



```
