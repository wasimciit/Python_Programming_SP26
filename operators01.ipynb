Q1. Write a Python program to take two numbers from the user and print:

Their sum

Difference

Product

Division

```python
num1=float(input("Enter the first number: "))
num2=float(input("Enter the second number: "))
sum=num1+num2
diff=num1-num2
pro=num1*num2
div=num1/num2
print("sum of numbers:",sum)
print("Differnece of the numbers:",diff)
print("Muliplication of numbers:",pro)
print('Division of numbers:',div)
```

Q2. Write a program to calculate the area and perimeter of a rectangle using user input for length and width.

```python
length=float(input("Enter the length of the rectangle: "))
width=float(input("Enter the width of the rectangle: "))
area=length*width
perimeter=2*(length+width)
print(f"Area of the rectangle:{area}")
print(f'perimeter of the rectangle:{perimeter}')
```

Q3. Write a program to calculate the square and cube of a given number.

```python
num=float(input("Enter the number: "))
square=num**2
cube=num**3
print(f"Square of the number:{square}")
print(f"Cube of the number:{cube}")
```

Q4. Write a program to check whether a number is even or odd using the modulus operator.

```python
num=int(input("Enter the number: "))
if num%2==0:
  print("The number is even")
else:
  print('The number is odd')
```

Q5. Write a program to calculate the power of a number

```python
base = float(input("Enter the base number: "))
exponent = float(input("Enter the exponent: "))
power = base ** exponent
print(f"The power of {base} to the {exponent} is: {power}")
```
