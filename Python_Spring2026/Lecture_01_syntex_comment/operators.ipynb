## Assignment Operators

- Assignment operators are used to assign values to variables:
  

```python
# Assignment operators table

print("{:<5} {:<5} {:<15} {:<20}".format("No", "Op", "Example", "Equivalent"))
print("-" * 55)

data = [
    (1, "=", "x = 5", "x = 5"),
    (2, "+=", "x += 3", "x = x + 3"),
    (3, "-=", "x -= 3", "x = x - 3"),
    (4, "*=", "x *= 3", "x = x * 3"),
    (5, "/=", "x /= 3", "x = x / 3"),
    (6, "%=", "x %= 3", "x = x % 3"),
    (7, "//=", "x //= 3", "x = x // 3"),
    (8, "**=", "x **= 3", "x = x ** 3"),
    (9, "&=", "x &= 3", "x = x & 3"),
    (10, "|=", "x |= 3", "x = x | 3"),
    (11, "^=", "x ^= 3", "x = x ^ 3"),
    (12, ">>=", "x >>= 3", "x = x >> 3"),
    (13, "<<=", "x <<= 3", "x = x << 3"),
    (14, ":=", "print(x := 3)", "x = 3; print(x)")
]

for row in data:
    print("{:<5} {:<5} {:<15} {:<20}".format(*row))
```
