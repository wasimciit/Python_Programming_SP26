# List

In Python, a List is the most versatile way to store a collection of data. You can think of it like a physical paper list where you can add items, cross them out, or change their order at any time.

Here are the three defining characteristics of a list:

1. Ordered: The items stay in the exact sequence you define. If "Apple" is first, it stays first unless you move it.

2. Mutable: This is a fancy way of saying "changeable." You can add new items, remove old ones, or update an existing item (like changing "Bread" to "Whole Wheat Bread").

3. Allows Duplicates: You can have the same value multiple times in one list

```python
# create list
fruits=['apple','kiwi','banana','mango','cherry']
```

```python
print(fruits)
```

```python
print(type(fruits))
```

```python
print(len(fruits))
```

```python
# access list items
print(fruits[1])
```

```python
fruits[:3]
```

```python
# check if item exist in the list
if 'strawberry' in fruits:
  print('Strawberry is in the list')
```

```python
if 'strawberry' not in fruits:
  print('Strawberry is not in the list')
```

```python
if 'apple' in fruits:
  print(' yes apple is in the fruits')
```

```python
# change item value by indexing
fruits[1]='strawberry'
print(fruits)
```

```python
# change a range of item values
fruits[0:2]=['kiwi','cherry']
print(fruits)
```

```python
# add list item (append and insert)
fruits.append('orange')
print(fruits)
```

```python
fruits.insert(2,'pineapple')
print(fruits)
```

```python
num=[10,20,30,40]
num.append(50)
print(num)
```

```python
num.insert(0,0)
print(num)
```

```python
# extend list
lis1=[1,3,5,7]
lis2=[2,4,6,8]
lis1.extend(lis2)
print(lis1)
```

```python
lis2.extend(lis1)
print(lis2)
```

```python
# remove list item (remove, pop, del, clear)
fruits.remove('kiwi')
print(fruits)
```

```python
lis3=[1,11,22,33,44,55]
lis3.pop(2)
print(lis3)
```

```python
lis3.pop()
print(lis3)
```

```python
lis3.clear()
print(lis3)
```

```python
# loop lists
fruits=['apple','kiwi','banana','mango','cherry']
for x in fruits:
  if 'app' in x:
    print(x)
```

```python
# looping through the index number
for i in range(len(fruits)):
  print(fruits[i])
```
