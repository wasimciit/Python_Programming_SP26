##**Arguments**
Information can be passed into functions as arguments.

Arguments are specified after the function name, inside the parentheses. You can add as many arguments as you want, just separate them with a comma.



```python
def my_function(first_name):
  print(first_name + " Raees")

my_function("Alexender")
my_function("Tariq")
my_function("Liyari")
```

###***Parameters vs Arguments***

*   A parameter is the variable listed inside the parentheses in the function definition.

* An argument is the actual value that is sent to the function when it is called.



```python
def class_fun(name): # name is a parameter
  print("Hello", name)

class_fun("Amal") # "Amal" is an argument
```

###**Number of Arguments**
* By default, a function must be called with the correct number of arguments.

* If your function expects 2 arguments, you must call it with exactly 2 arguments.

```python
def CR_name(f_name, l_name):
  print(f_name + " " + l_name)

CR_name("shadab", "Khan")
```

```python
# If you try to call the function with the wrong number of arguments, you will get an error:
def CR_name(f_name, l_name):
  print(f_name + " " + l_name)

CR_name("shadab")
```

###***Default Parameter Values***

You can assign default values to parameters. If the function is called without an argument, it uses the default value:

```python
def my_function(first_name="Ali"):
  print(first_name + " Raees")

my_function("Alexender")
my_function("Tariq")
my_function("Liyari")
my_function()
```

###***Keyword Arguments***

* You can send arguments with the key = value syntax.

```python
def my_pet(pet, name):
  print("I have a", pet)
  print("My", pet + "'s name is", name)

my_pet(pet = "parrot", name = "Ajwa")
```

```python
# This way, with keyword arguments, the order of the arguments does not matter.
def my_pet(pet, name):
  print("I have a", pet)
  print("My", pet + "'s name is", name)

my_pet(name = "Ajwa", pet = "parrot")
```

###**Positional Arguments**

* When you call a function with arguments without using keywords, they are called positional arguments.

* Positional arguments must be in the correct order:

```python
def my_pet(pet, name):
  print("I have a", pet)
  print("My", pet + "'s name is", name)

my_pet("parrot", "Ajwa")
```

```python
def my_pet(pet, name):
  print("I have a", pet)
  print("My", pet + "'s name is", name)

my_pet("Ajwa", "parrot")
```

###***Mixing Positional and Keyword Arguments***

* You can mix positional and keyword arguments in a function call.

* However, positional arguments must come before keyword arguments

```python
def my_pet(pet, name, age):
  print("I have a", age, "year old", pet, "named", name)

my_pet("parrot", age = 2, name = "Ajwa")
```

###***Passing Different Data Types***

* You can send any data type as an argument to a function (string, number, list, dictionary, etc.).

* The data type will be preserved inside the function:

```python
def fav(phaals):
  for phaal in phaals:
    print(phaal)

phaals = ["apple", "banana", "cherry"]
fav(phaals)
```

###***Returning Different Data Types***

* Functions can return any data type, including lists, tuples, dictionaries, and more.

```python
def fav():
  return ["apple", "banana", "cherry"]
fav()
```

```python
def my_function():
  return (10, 20)

x, y = my_function()
print("x:", x)
print("y:", y)
```

### *args and **kwargs
* By default, a function must be called with the correct number of arguments.

* However, sometimes you may not know how many arguments that will be passed into your function.

* *args and **kwargs allow functions to accept a unknown number of arguments.

```python
def my_function(*args):
  print("Type:", type(args))
  print("First argument:", args[0])
  print("Second argument:", args[1])
  print("All arguments:", args)

my_function("Ali", "Tayaab", "Lahori")
```

```python
def my_function(greeting, *names):
  for name in names:
    print(greeting, name)

my_function("Hello", "Ali", "Tayaab", "Lahori")
```

```python
def my_function(*numbers):
  total = 0
  for num in numbers:
    total += num
  return total

print(my_function(1, 2, 3))
print(my_function(10, 20, 30, 40))
print(my_function(5))
```

```python
def my_function(*numbers):
  if len(numbers) == 0:
    return None
  max_num = numbers[0]
  for num in numbers:
    if num > max_num:
      max_num = num
  return max_num

print(my_function(3, 7, 2, 9, 1))
```

```python
def my_function(**myvar):
  print("Type:", type(myvar))
  print("Name:", myvar["name"])
  print("Age:", myvar["age"])
  print("All data:", myvar)

my_function(name = "Tayyab", age = 30, city = "Pakistan")
```
